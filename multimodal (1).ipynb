```python
pdf = pdfplumber.open("NVDA-F4Q25-Quarterly-Presentation-FINAL.pdf")

print("Pages:", len(pdf.pages))
```

```python
documents = []

for i, page in enumerate(pdf.pages):

    text = page.extract_text()

    if text:

        documents.append({
            "page": i + 1,
            "text": text
        })

print("Extracted Pages:", len(documents))
```

```python
for doc in documents[:3]:

    print("=" * 60)
    print("Page:", doc["page"])
    print(doc["text"][:300])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)
```

```python
chunked_documents = []

chunk_id = 1

for doc in documents:

    chunks = splitter.split_text(doc["text"])

    for chunk in chunks:

        chunked_documents.append({
            "chunk_id": chunk_id,
            "page": doc["page"],
            "text": chunk
        })

        chunk_id += 1

print("Total Chunks:", len(chunked_documents))
```

```python
def retrieve_context(question, k=6):
    results = vector_db.similarity_search(question, k=k)

    context = "\n\n".join(
        [
            f"Page {doc.metadata['page']}:\n{doc.page_content}"
            for doc in results
        ]
    )

    return context, results
```

```python
def ask_financial_agent(question):
    context, results = retrieve_context(question)

    prompt = f"""
You are an expert financial analyst.

Use ONLY the information provided below.

If the answer is not available, say:
"I couldn't find that information in the report."

Context:
{context}

Question:
{question}

Requirements:
- Answer clearly.
- Use bullet points if appropriate.
- Mention the source page(s) at the end.
"""

    response = llm.invoke(prompt)

    return response.content
```

```python
from langgraph.graph import StateGraph
from langgraph.graph import END
```

```python
from typing import TypedDict, List, Any

class AgentState(TypedDict, total=False):
    question: str
    rewritten_question: str
    retrieved_docs: List[Any]
    context: str
    answer: str
    history: List[str]

print("AgentState defined successfully!")
```

```python
def generate_node(state):
    docs = state["retrieved_docs"]

    context = "\n\n".join(
        f"Page {doc.metadata.get('page')}:\n{doc.page_content}"
        for doc in docs
    )

    question = state.get("rewritten_question") or state["question"]

    prompt = f"""
You are an expert financial analyst.
Answer ONLY using the provided context.
Do not invent facts.
If the answer is missing, say:
"I couldn't find that information in the report."

Context:
{context}

Question:
{question}

Mention the source page(s).
"""

    response = llm.invoke(prompt)

    return {
        "context": context,
        "answer": response.content
    }
```

```python
def rewrite_node(state):

    history = "\n".join(state.get("history", []))

    prompt = f"""
You are a query rewriting assistant.

Rewrite the user's question into a complete and searchable financial question.

Conversation History:
{history}

Current Question:
{state["question"]}

Return ONLY the rewritten question.
"""

    response = llm.invoke(prompt)

    if isinstance(response.content, list):
        rewritten = "".join(
            part if isinstance(part, str)
            else getattr(part, "text", str(part))
            for part in response.content
        )
    else:
        rewritten = response.content

    return {
        "rewritten_question": rewritten.strip()
    }
```

```python
retriever = vector_db.as_retriever(
    search_kwargs={"k": 5}
)

print("Retriever ready!")
```

```python
def retrieve_node(state):
    question = state.get("rewritten_question") or state["question"]

    docs = retriever.invoke(question)

    print(f"Retrieved {len(docs)} documents")

    return {"retrieved_docs": docs}
```

```python
test_state = {
    "question": "What is Blackwell?",
    "rewritten_question": "What is NVIDIA Blackwell?"
}

test_result = retrieve_node(test_state)

print(test_result.keys())
print("Documents:", len(test_result["retrieved_docs"]))
```

```python
from langgraph.graph import StateGraph, END

graph = StateGraph(AgentState)

graph.add_node("retrieve", retrieve_node)
graph.add_node("generate", generate_node)

graph.set_entry_point("retrieve")
graph.add_edge("retrieve", "generate")
graph.add_edge("generate", END)

financial_agent = graph.compile()

print("Financial Analyst Agent Ready!")
```

```python
result = financial_agent.invoke({
    "question": "What is NVIDIA Blackwell?",
    "history": []
})

print("Answer:")
print(result["answer"])
```

```python

history = []

while True:
    question = input("\nYou: ")

    if question.lower().strip() == "exit":
        print("Goodbye!")
        break

    state = {
        "question": question,
        "history": history
    }

    result = financial_agent.invoke(state)

    print("\nAgent:")

    print(result.get("answer", "No answer was generated."))

    history.append(
        f"User: {question}\n"
        f"Assistant: {result.get('answer', '')}"
    )

    print("\n" + "=" * 80)
```
